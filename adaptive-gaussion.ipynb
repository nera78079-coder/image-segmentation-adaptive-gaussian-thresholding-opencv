```python
import cv2
import matplotlib.pyplot as plt
```

```python
img=cv2.imread("mountain.jpg")
img.shape
```

```python
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(img_gray,cmap="gray")
plt.axis('off')
```

```python
th_gauss = cv2.adaptiveThreshold(img_gray,255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY,199,5)
th_gauss.shape
```

```python
plt.imshow(th_gauss,cmap="gray")
plt.axis('off')
```
